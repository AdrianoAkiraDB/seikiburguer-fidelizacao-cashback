```python
import pandas as pd

# 1. Carregar os arquivos
df_supabase = pd.read_csv('clientes_supabase.csv')
df_telefones = pd.read_csv('telefones_original.csv')

# 2. Limpeza de nomes
df_supabase['nome'] = df_supabase['nome'].str.strip()
df_telefones['nome'] = df_telefones['nome'].str.strip()

# 3. O MATCH (Une as tabelas)
df_final = pd.merge(df_telefones, df_supabase[['id', 'nome']], on='nome', how='left')

# 4. REMOVER DUPLICADOS (A solução para o erro do Supabase)
# Isso garante que cada número apareça apenas UMA VEZ no arquivo final
df_final = df_final.drop_duplicates(subset=['telefone'])

# 5. Remove vazios
df_final = df_final.dropna(subset=['id', 'telefone'])

# 6. Formata para o Supabase
df_importar = df_final[['id', 'telefone']].rename(columns={
    'id': 'cliente_id',
    'telefone': 'numero'
})

# 7. Salva o arquivo pronto
df_importar.to_csv('importar_dim_telefones.csv', index=False)

print(f"✅ SUCESSO! Arquivo gerado com {len(df_importar)} linhas únicas.")
```
